```python
import telegram.ext
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
TOKEN= os.getenv("TOKEN")
print(TOKEN)
```

```python
import asyncio
import nest_asyncio
from telegram import Update
from telegram.ext import ApplicationBuilder, CommandHandler, ContextTypes

# Apply patch to allow nested event loops (needed in Jupyter/IDEs)
nest_asyncio.apply()

# Command handler for /start and /hello
async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user = update.effective_user
    await update.message.reply_text(f"Hello, {user.first_name}! Welcome to the bot.")
    await update.message.reply_text("Use /help to see available commands.")
# Function to safely run the bot
def run_bot():
    app = ApplicationBuilder().token(TOKEN).build()

    # Register command handlers
    app.add_handler(CommandHandler("start", start))
    app.add_handler(CommandHandler("hello", start))
    app.add_handler(CommandHandler("help", lambda update, context: update.message.reply_text("Available commands: /start, /hello,/ help, /message, /echo, /repeat, /info, /status, /time, /date, /location, /weather, /news, /joke, /quote, /fact, /advice.")))
    print("✅ Bot is running... Press Ctrl+C to stop.")
    app.add_handler(CommandHandler("message", handle_message))

    try:
        loop = asyncio.get_running_loop()
        loop.create_task(app.run_polling())
    except RuntimeError:
        app.run_polling()
def handle_message(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user = update.effective_user
    message = update.message.text
    response = f"You said: {message}. Hello, {user.first_name}!"
    context.bot.send_message(chat_id=update.effective_chat.id, text=response)
# Entry point
if __name__ == "__main__":
    run_bot()
```
