```python
required_nf_df=netflix_df[['Title','Description','Content Type','Genres']]
required_nf_df.head()
```

```python
#Drop the rows which has null values
required_nf_df.dropna()
```

```python
required_nf_df.count()
```

```python
#Clean the data
import nltk
nltk.download('stopwords')
```

```python
stemmer=nltk.SnowballStemmer('english')
from nltk.corpus import stopwords
import string
stopword=set(stopwords.words('english'))
```

```python
#Clean the data
import re
def clean(text):
    text = str(text).lower() # Convert to string and lowercase
    text = re.sub('\[.*?\]','', text) # Remove square brackets and their contents 
    text = re.sub('https?://\S+|www\,\S++', '', text)  # Remove URLs
    text = re.sub('<.*?>+', '', text)# Remove HTML tags
    text = re.sub('[%s]' % re.escape(string.punctuation), '', text)# Remove punctuation
    text = re.sub('\n', '', text)# Remove new line characters
    text = re.sub('\w*\d\w*', '', text)# Remove words containing numbers
    text = [word for word in text.split(' ') if word not in stopword]# Remove stopwords
    text = ' '.join(text)# Join words back into a single string
    text = [stemmer.stem(word) for word in text.split(' ')]# Stem the words
    text = ' '.join(text)# Join stemmed words back into a single string
    return text
```

```python
required_nf_df["Title"] = required_nf_df["Title"].apply(clean)
required_nf_df.head()
```

```python
required_nf_df.Title.sample(10)
```

```python
genreList = required_nf_df['Genres'].tolist()
tfidf=text.TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(genreList)
similarity= cosine_similarity(tfidf_matrix)
```

```python
indices=pd.Series(required_nf_df.index, index=required_nf_df['Title']).drop_duplicates()
```

```python
def netflix_recommendation(title, similarity=similarity):
    index = indices[title]
    similarity_scores = list(enumerate(similarity[index]))
    similarity_scores =similarity_scores[0:10] 
    movieindices = [i[0] for i in similarity_scores]
    return required_nf_df['Title'].iloc[movieindices]
```
