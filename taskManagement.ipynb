```python
def task():
    tasks=[] #empty list
    print("Welcome to the task management system")

    total_tasks = int(input("Enter the number of tasks you want to add: "))
    for i in range(1,total_tasks+1):
        task_name = input(f"Enter the name of task {i}: ")
        tasks.append(task_name)
    
    print(f"Here are your tasks:\n{tasks}")

    while True:
        operation=int(input("Enter 1-Add\n2-Update\n3-Delete\n4-View\n5-Exit/Stop/"))
        if operation == 1:
            add= input("Enter the task you want to add: ")
            tasks.append(add)
            print(f"Task {add} has been successfully added.")
        
        elif operation == 2:
            update = input("Enter the task you want to update: ")
            if update in tasks:
                new_task = input("Enter the new task: ")
                index = tasks.index(update)
                tasks[index] = new_task
                print(f"Task {update} has been updated to {new_task}.")

        elif operation == 3:
            delete = input("Enter the task you want to delete: ")
            if delete in tasks:
                tasks.remove(delete)
                print(f"Task {delete} has been deleted.")
            else:
                print(f"Task {delete} not found.")

        elif operation == 4:
            print(f"Here are your tasks:{tasks}")
        
        elif operation == 5:
            print("Exiting the task management system. Goodbye!")
            break

        else:
            print("Invalid operation. Please try again.")
task()
            
```
